```python
from pathlib import Path
import shutil

src = Path("/content/ECERC/MELD/results")
dst = Path("/content/drive/MyDrive/ECERC/results/stage1")

dst.mkdir(parents=True, exist_ok=True)

stage1_runs = [
    "CB0.999_s0",
    "CB0.9999_s0",
    "FL2_s0",
    "OS2_s0",
    "OS4_s0",
]

for run in stage1_runs:
    source = src / run
    target = dst / run

    if target.exists():
        shutil.rmtree(target)

    shutil.copytree(source, target)
    print(f"Saved: {run}")

# Save the summary and record
for name in ["stage1_record.txt"]:
    source = src / name
    if source.exists():
        shutil.copy2(source, dst / name)
        print(f"Saved: {name}")

# Save analyzer-generated summary directory
summary_src = src / "stage1_summary"
summary_dst = dst / "stage1_summary"

if summary_src.exists():
    if summary_dst.exists():
        shutil.rmtree(summary_dst)
    shutil.copytree(summary_src, summary_dst)
    print("Saved: stage1_summary")
```
