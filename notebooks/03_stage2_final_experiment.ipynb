```python
from pathlib import Path
import shutil

src = Path("/content/ECERC/MELD/results")
dst = Path("/content/drive/MyDrive/ECERC/results/stage2")
dst.mkdir(parents=True, exist_ok=True)

runs = ["B0_s0", "B0_s1", "B0_s2"]

for run in runs:
    source = src / run
    target = dst / run
    if source.exists():
        if target.exists():
            shutil.rmtree(target)
        shutil.copytree(source, target)
        print(f"Saved: {run}")
```
